## Exercise 6.1.1

To meassure the local norm of the tangent vector $\mathbf{v}_1 = (1, 0)^\top$ using the function:
$$
\mathbf{G}_x = (1 + \|x\|^2) \mathbf{I}
$$
we can compute the local norm as follows:
$$
\|\mathbf{v}_1\|_x = \sqrt{\mathbf{v}_1^\top \mathbf{G}_x \mathbf{v}_1} = \sqrt{\mathbf{v}_1^\top (1 + \|x\|^2) \mathbf{I} \mathbf{v}_1} = \sqrt{(1 + \|x\|^2) \mathbf{v}_1^\top \mathbf{v}_1}
$$


We can do this for the 3 x points

$$
\mathbf{x}_1 = (1, 1)^\top, \quad \mathbf{x}_2 = (2, 3)^\top, \quad \mathbf{x}_3 = (0, 3)^\top
$$
Calculating the local norm for each point:
1. For $\mathbf{x}_1$:
$$\|\mathbf{v}_1\|_{x_1} = \sqrt{(1 + \|x_1\|^2) \mathbf{v}_1^\top \mathbf{v}_1} = \sqrt{(1 + \sqrt{1^2 + 1^2}^2) \cdot (1, 0)^\top (1, 0)} = \sqrt{(1 + 2) \cdot 1} = \sqrt{3}$$
2. For $\mathbf{x}_2$:
$$\|\mathbf{v}_1\|_{x_2} = \sqrt{(1 + \|x_2\|^2) \mathbf{v}_1^\top \mathbf{v}_1} = \sqrt{(1 + \sqrt{2^2 + 3^2}^2) \cdot (1, 0)^\top (1, 0)} = \sqrt{(1 + 13) \cdot 1} = \sqrt{14}$$
3. For $\mathbf{x}_3$:
$$\|\mathbf{v}_1\|_{x_3} = \sqrt{(1 + \|x_3\|^2) \mathbf{v}_1^\top \mathbf{v}_1} = \sqrt{(1 + \sqrt{0^2 + 3^2}^2) \cdot (1, 0)^\top (1, 0)} = \sqrt{(1 + 9) \cdot 1} = \sqrt{10}$$

$$

$$
We have $\mathbf{v}_1 = (1, 0)^\top$ and $\mathbf{v}_2 = (0, 1)^\top$. We can compute the local angles for each point:

We have the geodesic ODE given my the two equations:
$$
[\ddot{c}_t]_m = -\sum_{j=1}^d \sum_{k=1}^d \Gamma_{jk}^m(c_t) [\dot{c}_t]_j [\dot{c}_t]_k
$$
and
$$
\Gamma_{jk}^m(x) = \frac{1}{2} \sum_{i=1}^d [\mathbf{G}_x^{-1}]_{mi} \left( 2 \frac{\partial [\mathbf{G}_x]_{ij}}{\partial [c_t]_k} - \frac{\partial [\mathbf{G}_x]_{jk}}{\partial [c_t]_i} \right)
$$

Using $\mathbf{G} = \mathbf{I}$, we can compute the Christoffel symbols $\Gamma_{jk}^m$ as follows:
1. Compute $\mathbf{G}_x^{-1}$:
$$\mathbf{G}_x^{-1} = \mathbf{I}^{-1} = \mathbf{I}$$
2. Compute the partial derivatives of $\mathbf{G}_x$:
$$\frac{\partial [\mathbf{G}_x]_{ij}}{\partial [c_t]_k} = \frac{\partial [\mathbf{I}]_{ij}}{\partial [c_t]_k} = 0$$
3. Substitute into the formula for $\Gamma_{jk}^m$:
$$\Gamma_{jk}^m(x) = \frac{1}{2} \sum_{i=1}^d [\mathbf{I}]_{mi} \left( 2 \cdot 0 - 0 \right) = 0$$

## Exercise 6.2.2

We can then substitute $\Gamma_{jk}^m = 0$ into the geodesic ODE:
$$[\ddot{c}_t]_m = -\sum_{j=1}^d \sum_{k=1}^d 0 \cdot [\dot{c}_t]_j [\dot{c}_t]_k = 0$$
This implies that $\ddot{c}_t = 0$, which means that the geodesics are straight lines in this case. Therefore, the geodesic ODE simplifies to:
$$\ddot{c}_t = 0$$
This indicates that the geodesics in this space are straight lines, which is consistent with the fact that we are working with the identity metric $\mathbf{G} = \mathbf{I}$.

We consider a geodesic with initial conditions: $c_0 = \mathbf{0}$, and $\dot{c}_0 = \mathbf{v}$, where $\mathbf{v}$ is a tangent vector at the origin.

Then the acceleration at $\ddot{c}_0$ meaning $t=0$ can be computed using the geodesic ODE:
$$[\ddot{c}_0]_m = -\sum_{j=1}^d \sum_{k=1}^d \Gamma_{jk}^m(c_0) [\dot{c}_0]_j [\dot{c}_0]_k$$

Inserting the Christoffel symbols at $c_0 = \mathbf{0}$, we have:
$$\Gamma_{jk}^m(c_0) = \frac{1}{1 + \|c_0\|^2} \left( 2 [c_0]_k \mathbf{I}_{mj} - [c_0]_m \mathbf{I}_{jk} \right) = 0$$

This implies that $\ddot{c}_0 = 0$, which means that the geodesic has zero acceleration at the origin. Therefore, the geodesic that starts at the origin with an initial velocity $\mathbf{v}$ will be a straight line in the direction of $\mathbf{v}$, at least locally around the origin.
