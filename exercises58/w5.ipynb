## 5.1.1 Distance is the same

We want to show that:
$$\| x_i - x_j \| = \| \hat{x}_i - \hat{x}_j \|$$
We know that $\hat{x}_i = Rx_i$ and $\hat{x}_j = Rx_j$ where $R$ is the rotation matrix. We can then write:
$$x_i - x_j = R x_i - R x_j$$
Since $R$ is a linear transformation, we can write:
$$x_i - x_j = R (x_i - x_j) $$

We can then take the euclidean norm of both sides:
$$\| x_i - x_j \| = \| R (x_i - x_j) \|$$
Now we can square both sides:
$$\| x_i - x_j \|^2 = \| R (x_i - x_j) \|^2$$
We can then write the right hand side as:
$$\| R (x_i - x_j) \|^2 = (R (x_i - x_j))^T (R (x_i - x_j))$$
Since $R$ is an orthogonal matrix, we can write:
$$\| R (x_i - x_j) \|^2 = (x_i - x_j)^T R^T R (x_i - x_j)$$
Since $R^T R = I$, we can write:
$$\| R (x_i - x_j) \|^2 = (x_i - x_j)^T (x_i - x_j)$$
We can then take the square root of both sides:
$$\| R (x_i - x_j) \| = \| x_i - x_j \|$$
We can then write:
$$\| \hat{x}_i - \hat{x}_j \| = \| R (x_i - x_j) \| = \| x_i - x_j \|$$

## 5.1.2 Angles are the same

The angle between two vectors $x_i - x_j$ and $x_k - x_j$ are equal, we can show this by:
$$\angle (x_i - x_j, x_k - x_j) = \angle (R (x_i - x_j), R (x_k - x_j))$$
We can then write the left hand side as:
$$\angle (x_i - x_j, x_k - x_j) = \frac{(x_i - x_j)^T (x_k - x_j)}{\| x_i - x_j \| \| x_k - x_j \|}$$
We can then write the right hand side as:
$$\angle (R (x_i - x_j), R (x_k - x_j)) = \frac{(R (x_i - x_j))^T (R (x_k - x_j))}{\| R (x_i - x_j) \| \| R (x_k - x_j) \|}$$
From 5.1.1 we know that $\| R (x_i - x_j) \| = \| x_i - x_j \|$ and $\| R (x_k - x_j) \| = \| x_k - x_j \|$, we can then write:
$$\angle (R (x_i - x_j), R (x_k - x_j)) = \frac{(R (x_i - x_j))^T (R (x_k - x_j))}{\| x_i - x_j \| \| x_k - x_j \|}$$
Since $R$ is an orthogonal matrix, we can write:
$$\angle (R (x_i - x_j), R (x_k - x_j)) = \frac{(x_i - x_j)^T R^T R (x_k - x_j)}{\| x_i - x_j \| \| x_k - x_j \|}$$
Since $R^T R = I$, we can write:
$$\angle (R (x_i - x_j), R (x_k - x_j)) = \frac{(x_i - x_j)^T (x_k - x_j)}{\| x_i - x_j \| \| x_k - x_j \|}$$
We can then write:
$$\angle (R (x_i - x_j), R (x_k - x_j)) = \angle (x_i - x_j, x_k - x_j)$$

# Opgave 5.2

Consider a generative model

$$
\mathbf{y} = f(\mathbf{x}),
$$

where $\mathbf{x} = (x_1, x_2) \in \mathbb{R}^2$ and $\mathbf{y} = (y_1, y_2, y_3) \in \mathbb{R}^3$.

Let

$$
f(\mathbf{x}) =
\begin{pmatrix}
f_1(\mathbf{x}) \\
f_2(\mathbf{x}) \\
f_3(\mathbf{x})
\end{pmatrix}
=
\begin{pmatrix}
2x_1^2 + x_2^2 \\
x_1 \\
x_2
\end{pmatrix}.
$$

## 5.2.1 Jacobian of the generative model

We now want to find the Jacobian matrix of $f$ at $\mathbf{x}$. The Jacobian matrix is given by:
$$
J_f(\mathbf{x}) =
\begin{pmatrix}
\frac{\partial f_1}{\partial x_1} & \frac{\partial f_1}{\partial x_2} \\
\frac{\partial f_2}{\partial x_1} & \frac{\partial f_2}{\partial x_2} \\
\frac{\partial f_3}{\partial x_1} & \frac{\partial f_3}{\partial x_2}
\end{pmatrix}.
$$
We can then compute the partial derivatives:
$$
\frac{\partial f_1}{\partial x_1} = 4x_1, \quad \frac{\partial f_1}{\partial x_2} = 2x_2, \quad \frac{\partial f_2}{\partial x_1} = 1, \quad \frac{\partial f_2}{\partial x_2} = 0, \quad \frac{\partial f_3}{\partial x_1} = 0, \quad \frac{\partial f_3}{\partial x_2} = 1.
$$
We can then write the Jacobian matrix as:
$$
J_f(\mathbf{x}) =
\begin{pmatrix}
4x_1 & 2x_2 \\
1 & 0 \\
0 & 1
\end{pmatrix}.
$$

```python
# Write up model with sympy
import sympy as sp
x1, x2 = sp.symbols('x1 x2')
f1 = 2*x1**2 + x2**2
f2 = x1
f3 = x2
f = sp.Matrix([f1, f2, f3])
J_f = f.jacobian([x1, x2])
J_f
```

## 5.2.2 Show immersed manifold

To show that the generative models spans an immersed manifold, we need to show that the Jacobian matrix has full rank. 

We can do this for a 3x2 matrix by showing the reduced row echelon form of the Jacobian matrix has 2 leading ones since the number of columns is 2.

We can then write the Jacobian matrix as:
$$J_f(\mathbf{x}) =
\begin{pmatrix}
4x_1 & 2x_2 \\
1 & 0 \\
0 & 1
\end{pmatrix}.
$$
We can then perform row operations to get the reduced row echelon form:
1. Swap row 1 and row 2:
$$\begin{pmatrix}
1 & 0 \\
4x_1 & 2x_2 \\
0 & 1
\end{pmatrix}$$
2. Subtract 4x_1 times row 1 from row 2:
$$\begin{pmatrix}
1 & 0 \\
0 & 2x_2 \\
0 & 1
\end{pmatrix}$$
3. Divide row 2 by 2x_2:
$$\begin{pmatrix}
1 & 0 \\
0 & 1 \\
0 & 1
\end{pmatrix}$$
4. Subtract row 2 from row 3:
$$\begin{pmatrix}
1 & 0 \\
0 & 1 \\
0 & 0
\end{pmatrix}$$
Since the reduced row echelon form has 2 leading ones, we can conclude that the Jacobian matrix has full rank and therefore the generative model spans an immersed manifold.

```python
# Find the rref of the Jacobian
rref_J_f = J_f.rref()[0]
rref_J_f
```

```python
# Compute rank of the Jacobian
rank_J_f = J_f.rank()
rank_J_f
```

## 5.2.3 Show embedded manifold

Now we want to show that the generative model spans an embedded manifold. 

We can show this by showing that the left inverse of the Jacobian matrix exists. The left inverse of a matrix $A$ is given by:

$$A^{-1} = (A^T A)^{-1} A^T.$$

We can then write the Jacobian matrix as:
$$J_f(\mathbf{x}) =
\begin{pmatrix}
4x_1 & 2x_2 \\
1 & 0 \\
0 & 1
\end{pmatrix}.
$$
We can then compute the left inverse of the Jacobian matrix:
$$J_f^{-1} = (J_f^T J_f)^{-1} J_f^T.$$
We can then compute $J_f^T J_f$:
$$J_f^T J_f =
\begin{pmatrix}
16x_1^2 + 1 & 8x_1 x_2 \\
8x_1 x_2 & 4x_2^2 + 1
\end{pmatrix}.$$
We can then compute the inverse of $J_f^T J_f$:
$$(J_f^T J_f)^{-1} =
\frac{1}{(16x_1^2 + 1)(4x_2^2 + 1) - (8x_1 x_2)^2}
\begin{pmatrix}
4x_2^2 + 1 & -8x_1 x_2 \\
-8x_1 x_2 & 16x_1^2 + 1
\end{pmatrix}.$$
Reducing the expression we get:
$$(J_f^T J_f)^{-1} =
\frac{1}{16x_1^2 + 4x_2^2 + 1}
\begin{pmatrix}
4x_2^2 + 1 & -8x_1 x_2 \\
-8x_1 x_2 & 16x_1^2 + 1
\end{pmatrix}.$$
We know that this can not be zero since $16x_1^2 + 4x_2^2 + 1 > 0$ for all $x_1, x_2$. 

Therefore, the left inverse of the Jacobian matrix exists and we can conclude that the generative model spans an embedded manifold.
